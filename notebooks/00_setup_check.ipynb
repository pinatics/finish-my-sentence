```python
import sys
import importlib.util

print("Python:", sys.version)

packages = [
    "torch",
    "transformers",
    "accelerate",
    "nnsight",
    "pandas",
    "numpy",
    "matplotlib",
    "scipy",
]

for pkg in packages:
    found = importlib.util.find_spec(pkg) is not None
    print(f"{pkg:12} -> {'FOUND' if found else 'MISSING'}")
```
