# 02 Dataset Design

Build 15 structural minimal pairs across five relation types.

The goal is to keep the lexical material as similar as possible while changing the relation itself, so later measurements are less likely to succeed from word presence alone.

Pair 1 is reserved for pilot development. The remaining 14 pairs are held out.

## Tokenizer check

Use the same Qwen tokenizer as the main experiment to check token length and token identity.

```python
from transformers import AutoTokenizer

MODEL = "Qwen/Qwen3.5-4B"
tok = AutoTokenizer.from_pretrained(MODEL)

print("Tokenizer loaded:", type(tok).__name__)
```

## Pair 1: role / argument binding

`The user tests the model.`  
`The model tests the user.`

Check that the lexical material is matched while the semantic roles reverse.

```python
pair_a = "The user tests the model."
pair_b = "The model tests the user."

for text in [pair_a, pair_b]:
    ids = tok.encode(text, add_special_tokens=False)
    tokens = [tok.decode([i]) for i in ids]

    print(text)
    print("IDs:", ids)
    print("Tokens:", tokens)
    print("Length:", len(ids))
    print()
```

## Validation criteria

For each pair I check:

- matched token length
- matched token multiset where possible
- clean structural contrast
- no obvious lexical shortcut

```python
pairs = [
    ("Pair 1 — Role Binding", pair_a, pair_b),
    ("Pair 2 — Argument Direction", pair2_a, pair2_b),
    ("Pair 3 — Modifier Attachment", pair3_a, pair3_b),
    ("Pair 4 — Spatial Relation Direction", pair4_a, pair4_b),
    ("Pair 5 — Comparative Relation Direction", pair5_a, pair5_b),
]

for name, a, b in pairs:
    ids_a = tok.encode(a, add_special_tokens=False)
    ids_b = tok.encode(b, add_special_tokens=False)

    same_length = len(ids_a) == len(ids_b)
    same_multiset = sorted(ids_a) == sorted(ids_b)

    print(name)
    print("A:", a)
    print("B:", b)
    print("Same length:", same_length)
    print("Same token multiset:", same_multiset)
    print("-" * 60)
```

## Pair 13: initial quantifier version

`All dogs chase some cats.`  
`Some dogs chase all cats.`

This version matched in length but failed the identical token-multiset check because `All` and `Some` appeared in different tokenization contexts.

```python
pair13_a = "All dogs chase some cats."
pair13_b = "Some dogs chase all cats."

for text in [pair13_a, pair13_b]:
    ids = tok.encode(text, add_special_tokens=False)
    tokens = [tok.decode([i]) for i in ids]

    print(text)
    print("IDs:", ids)
    print("Tokens:", tokens)
    print("Length:", len(ids))
    print()
```

## Pair 13: quantifier binding

Add a shared prefix so both quantifiers appear in the same tokenization context.

`In this scene, all dogs chase some cats.`  
`In this scene, some dogs chase all cats.`

```python
pair13_a = "In this scene, all dogs chase some cats."
pair13_b = "In this scene, some dogs chase all cats."

for text in [pair13_a, pair13_b]:
    ids = tok.encode(text, add_special_tokens=False)
    tokens = [tok.decode([i]) for i in ids]

    print(text)
    print("IDs:", ids)
    print("Tokens:", tokens)
    print("Length:", len(ids))
    print()

ids_a = tok.encode(pair13_a, add_special_tokens=False)
ids_b = tok.encode(pair13_b, add_special_tokens=False)

print("Same length:", len(ids_a) == len(ids_b))
print("Same token multiset:", sorted(ids_a) == sorted(ids_b))
```

## Final validation

Check all 15 pairs for matched token length, matched token multiset, category labels, and structural consistency.

```python
validation_results = []

for item in final_dataset:
    ids_a = tok.encode(item["a"], add_special_tokens=False)
    ids_b = tok.encode(item["b"], add_special_tokens=False)

    same_length = len(ids_a) == len(ids_b)
    same_multiset = sorted(ids_a) == sorted(ids_b)

    validation_results.append({
        "pair_id": item["pair_id"],
        "category": item["category"],
        "len_a": len(ids_a),
        "len_b": len(ids_b),
        "same_length": same_length,
        "same_token_multiset": same_multiset,
        "valid": same_length and same_multiset,
    })

for r in validation_results:
    print(
        f"Pair {r['pair_id']:>2} | "
        f"{r['category']:<22} | "
        f"length={r['len_a']}/{r['len_b']} | "
        f"same_length={r['same_length']} | "
        f"same_multiset={r['same_token_multiset']} | "
        f"VALID={r['valid']}"
    )

print()
print("Total pairs:", len(validation_results))
print("All pairs valid:", all(r["valid"] for r in validation_results))
```
